```python
df_2021_path = data_dir / "ttc-subway-delay-data-2021.xlsx"
sheets_dict_21 = pl.read_excel(df_2021_path, sheet_id=0)
print("Sheet names:", sheets_dict_21.keys())
```

```python
def check_months_columns(df_dict):
    cols_dict = {}
    for sheet_name, df in df_dict.items():
        # 1. Check for mismatched datatypes
        for col in df.columns:
            if col in cols_dict:
                if df[col].dtype != cols_dict[col]:
                    print(f"⚠️ TYPE MISMATCH: {sheet_name} '{col}' is {df[col].dtype}, but was previously {cols_dict[col]}")
                    return
            else:
                # Record the datatype the first time we see a column
                cols_dict[col] = df[col].dtype
                
        # 2. Check if the current sheet is missing any previously seen columns
        for known_col in cols_dict:
            if known_col not in df.columns:
                print(f"⚠️ MISSING COLUMN: {sheet_name} is missing '{known_col}'")
                return
    print("all columns match")
```

```python
check_months_columns(sheets_dict_21)
```
